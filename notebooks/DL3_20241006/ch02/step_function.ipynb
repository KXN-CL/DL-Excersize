```python
# coding: utf-8
# 계단 함수 (Step Function) 구현
# 계단 함수는 입력이 0보다 크면 1을, 0보다 작거나 같으면 0을 출력하는 기본적인 활성화 함수입니다.
# 신경망에서 가장 단순한 비선형 활성화 함수 중 하나입니다.

import numpy as np
import matplotlib.pylab as plt


def step_function(x):
    """
    계단 함수 (Step Function) 활성화 함수.
    
    이 함수는 입력 x가 0보다 크면 1을 반환하고,
    0보다 작거나 같으면 0을 반환합니다.
    
    Args:
        x: 입력 값 - 스칼라, 리스트, 또는 numpy 배열 가능
        
    Returns:
        np.ndarray: 각 입력이 양수였는지 나타내는 0과 1로 구성된 배열
    """
    # numpy 배열로 변환하고 정수 타입 (dtype) 으로 설정
    # 비교 연산 (x > 0) 은 불리언 배열을 반환하고,
    # 이것이 정수로 변환됩니다 (True -> 1, False -> 0)
    return np.array(x > 0, dtype=int)


# -5.0 에서 5.0 까지 0.1 간격으로 입력 값 생성
# 계단 함수를 테스트할 값의 범위를 만듭니다
X = np.arange(-5.0, 5.0, 0.1)

# 모든 입력 값에 계단 함수를 적용합니다
# 0 보다 작거나 같은 값은 0 이 되고, 0 보다 큰 값은 1 이 됩니다
Y = step_function(X)

# 계단 함수를 그래프로 plotting 합니다
# X 는 입력 값을 포함하고, Y 는 해당하는 출력 값을 포함합니다
plt.plot(X, Y)

# 그래프 설정
plt.title('Step Function', fontsize=16)
plt.xlabel('Input (x)', fontsize=12)
plt.ylabel('Output (step_function(x))', fontsize=12)
plt.grid(True, alpha=0.3)

# y 축 범위를 설정하여 명확한 시각화를 제공합니다
# 계단 함수는 0 또는 1 만 출력하므로, 이 범위를 약간 넘도록 설정합니다
plt.ylim(-0.1, 1.1)  # y 축의 범위 설정

# x = 0 위치에 점선 표시 (계단 함수가 변화하는 지점)
plt.axvline(x=0, color='red', linestyle='--', alpha=0.5)

# 그래프를 표시합니다
plt.show()
```
