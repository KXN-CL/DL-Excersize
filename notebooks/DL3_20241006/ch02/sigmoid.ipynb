```python
# coding: utf-8
# 시그모이드 함수 (Sigmoid Function) 구현
# 시그모이드 함수는 신경망에서 널리 사용되는 활성화 함수로,
# 입력을 0과 1 사이의 값으로 변환하는 S자 형태의 함수입니다.
# 수식: σ(x) = 1 / (1 + exp(-x))

import numpy as np
import matplotlib.pylab as plt


def sigmoid(x):
    """
    시그모이드 함수 (Sigmoid Function) 활성화 함수.
    
    이 함수는 입력 x를 0과 1 사이의 값으로 변환합니다.
    큰 양수 입력은 1에 가깝게, 큰 음수 입력은 0에 가깝게 변환됩니다.
    
    Args:
        x: 입력 값 - 스칼라, 리스트, 또는 numpy 배열 가능
        
    Returns:
        np.ndarray: 시그모이드 함수를 적용한 출력 값 (0과 1 사이)
    """
    # 시그모이드 함수: 1 / (1 + exp(-x))
    # numpy의 exp 함수는 자연상수 e의 거듭제곱을 계산합니다
    return 1 / (1 + np.exp(-x))


# -7.0 에서 7.0 까지 0.1 간격으로 입력 값 생성
# 시그모이드 함수의 S자 형태를 명확히 보기 위한 범위입니다
X = np.arange(-7.0, 7.0, 0.1)

# 모든 입력 값에 시그모이드 함수를 적용합니다
Y = sigmoid(X)

# 시그모이드 함수를 그래프로 plotting 합니다
# X 는 입력 값을 포함하고, Y 는 해당하는 출력 값을 포함합니다
plt.plot(X, Y)

# 그래프 설정
plt.title('Sigmoid Function', fontsize=16)
plt.xlabel('Input (x)', fontsize=12)
plt.ylabel('Output (sigmoid(x))', fontsize=12)
plt.grid(True, alpha=0.3)

# y 축 범위를 설정합니다
# 시그모이드 함수는 0과 1 사이의 값을 출력합니다
plt.ylim(-0.1, 1.1)  # y 축의 범위 설정

# x = 0 위치에 점선 표시 (시그모이드 함수의 중심점, 출력 = 0.5)
plt.axvline(x=0, color='red', linestyle='--', alpha=0.5)
# y = 0.5 위치에 점선 표시
plt.axhline(y=0.5, color='green', linestyle='--', alpha=0.5)

# 그래프를 표시합니다
plt.show()
```
