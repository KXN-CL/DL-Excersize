# 다층 퍼셉트론 (Multi-Layer Perceptron, MLP)

다층 퍼셉트론은 입력층, 하나 이상의 은닉층, 출력층으로 구성된 신경망입니다.
각 층은 가중치(W)와 편향(B)을 가지며, 활성화 함수를 통해 비선형 변환을 수행합니다.

## 신경망 구조

```
입력층 (2개)     은닉층 1 (3개)     은닉층 2 (2개)     출력층 (2개)
   x1  ────►      z11  ────►          z21  ────►          y1
   x2  ────►      z12  ────►          z22  ────►          y2
                  z13  ────►
```

- **입력층**: 2개 뉴런 (입력 데이터: [1.0, 0.5])
- **은닉층 1**: 3개 뉴런 (활성화 함수: sigmoid)
- **은닉층 2**: 2개 뉴런 (활성화 함수: sigmoid)
- **출력층**: 2개 뉴런 (활성화 함수: softmax)

````python
# 필요한 라이브러리 Import
import numpy as np
import os, sys

# common.functions 모듈에서 활성화 함수 가져오기
# 현재 디렉토리에서 시작하여 상위 디렉토리의 common 모듈을 찾습니다
current_dir = os.getcwd()
parent_dir = os.path.dirname(current_dir)
sys.path.append(parent_dir)
from common.functions import sigmoid, softmax
````

---

## 1. 입력층에서 1층 (은닉층 1) 으로 신호 전달

입력 데이터 X에 가중치 W1을 곱하고 편향 B1을 더합니다.

**수식:** A1 = X × W1 + B1

| 변수 | 형태 | 설명 |
|------|------|------|
| X | (2,) | 입력 데이터 [1.0, 0.5] |
| W1 | (2, 3) | 입력층 → 은닉층 1 가중치 |
| B1 | (3,) | 은닉층 1 편향 |
| A1 | (3,) | 은닉층 1의 가중합 (미활성화 출력) |

````python
# 입력 데이터 정의
# 입력층의 2개 뉴런에 들어갈 값
X = np.array([1.0, 0.5])

# 은닉층 1의 가중치 (W1)
# 형태: (입력 뉴런 수, 은닉 뉴런 수) = (2, 3)
# W1[0, :] = X[0]에서 은닉층 뉴런으로 가는 가중치
# W1[1, :] = X[1]에서 은닉층 뉴런으로 가는 가중치
W1 = np.array([[0.1, 0.3, 0.5],
               [0.2, 0.4, 0.6]])

# 은닉층 1의 편향 (B1)
# 은닉층 1의 3개 뉴런 각각에 추가되는 값
B1 = np.array([0.1, 0.2, 0.3])

# 값 확인
print('X = ', X)
print('B1 = ', B1)
print('W1 =', W1)

# 가중합 계산: A1 = X * W1 + B1
# np.dot(X, W1) 은 벡터 X(2,) 와 행렬 W1(2,3) 의 행렬 곱셈으로 결과 형태 (3,)
# 여기에 편향 B1(3,) 을 더합니다 (브로드캐스팅)
A1 = np.dot(X, W1) + B1
print('X * W1 + B1 =', A1)
````

---

## 2. 활성화 함수의 적용 (은닉층 1)

가중합 A1에 시그모이드 활성화 함수를 적용하여 비선형 변환을 수행합니다.

**수식:** Z1 = sigmoid(A1) = 1 / (1 + exp(-A1))

시그모이드 함수는 모든 입력을 0과 1 사이로 압축하므로,
은닉층의 출력이 일정한 범위로 제한됩니다.

````python
# 시그모이드 활성화 함수 적용
# A1의 각 요소에 sigmoid 함수를 적용합니다
Z1 = sigmoid(A1)

# 결과 확인
print('A1 = ', A1)
print('Z1 = ', Z1)

# Z1이 은닉층 1의 출력이며, 다음 층으로 전달됩니다
# 형태: (3,) - 3개 뉴런의 출력
````

---

## 3. 1층 (은닉층 1) 에서 2층 (은닉층 2) 으로 신호 전달

은닉층 1의 출력 Z1에 가중치 W2를 곱하고 편향 B2를 더합니다.

**수식:** A2 = Z1 × W2 + B2

| 변수 | 형태 | 설명 |
|------|------|------|
| Z1 | (3,) | 은닉층 1 출력 |
| W2 | (3, 2) | 은닉층 1 → 은닉층 2 가중치 |
| B2 | (2,) | 은닉층 2 편향 |
| A2 | (2,) | 은닉층 2의 가중합 (미활성화 출력) |

````python
# 은닉층 2의 가중치 (W2)
# 형태: (은닉층 1 뉴런 수, 은닉층 2 뉴런 수) = (3, 2)
W2 = np.array([[0.1, 0.4],
               [0.2, 0.5],
               [0.3, 0.6]])

# 은닉층 2의 편향 (B2)
B2 = np.array([0.1, 0.2])

# 값 확인
print('B2 = ', B2)
print('W2 =', W2)

# 가중합 계산: A2 = Z1 * W2 + B2
A2 = np.dot(Z1, W2) + B2
print('Z1 * W2 + B2 =', A2)
````

---

## 4. 활성화 함수의 적용 (은닉층 2)

가중합 A2에 시그모이드 활성화 함수를 적용합니다.

**수식:** Z2 = sigmoid(A2)

````python
# 시그모이드 활성화 함수 적용
Z2 = sigmoid(A2)

# 결과 확인
print('A2 = ', A2)
print('Z2 = ', Z2)

# Z2가 은닉층 2의 출력이며, 출력층으로 전달됩니다
````

---

## 5. 2층 (은닉층 2) 에서 출력층으로 신호 전달

은닉층 2의 출력 Z2에 가중치 W3을 곱하고 편향 B3을 더합니다.

**수식:** A3 = Z2 × W3 + B3

| 변수 | 형태 | 설명 |
|------|------|------|
| Z2 | (2,) | 은닉층 2 출력 |
| W3 | (2, 2) | 은닉층 2 → 출력층 가중치 |
| B3 | (2,) | 출력층 편향 |
| A3 | (2,) | 출력층의 가중합 (미활성화 출력) |

````python
# 출력층의 가중치 (W3)
# 형태: (은닉층 2 뉴런 수, 출력 뉴런 수) = (2, 2)
W3 = np.array([[0.2, 0.5],
               [0.3, 0.6]])

# 출력층의 편향 (B3)
B3 = np.array([0.1, 0.2])

# 값 확인
print('B3 = ', B3)
print('W3 =', W3)

# 가중합 계산: A3 = Z2 * W3 + B3
A3 = np.dot(Z2, W3) + B3
print('Z2 * W3 + B3 =', A3)
````

---

## 6. 출력층에 Softmax 활성화 함수 적용

출력층에서는 Softmax 함수를 적용하여 출력을 확률 분포로 변환합니다.

**수식:** Y = softmax(A3) = exp(A3) / sum(exp(A3))

Softmax 함수의 특징:
- 모든 출력의 합이 1이 됩니다
- 각 출력은 0과 1 사이의 값을 가지며, 분류 확률로 해석할 수 있습니다

````python
# Softmax 활성화 함수 적용
# A3의 값을 확률로 변환합니다
Y = softmax(A3)

# 결과 확인
print('A3 = ', A3)
print('Y = ', Y)

# 출력값의 합이 1인지 확인
print(f'출력값 합계: {np.sum(Y)}')

# 각 클래스에 대한 예측 확률
print(f'클래스 0일 확률: {Y[0]:.4f}')
print(f'클래스 1일 확률: {Y[1]:.4f}')
````

## 전체 순전파 (Forward Propagation) 함수로 통합

위에서 단계별로 수행한 계산을 하나의 함수로 통합합니다.
이렇게 하면 새로운 입력 데이터에 대해 쉽게 추론이 가능합니다.

````python
# 다층 퍼셉트론 초기화
def init_network():
    """
    신경망의 가중치와 편향을 초기화합니다.
    
    Returns
    -------
    dict: 가중치(W)와 편향(B)을 키로 갖는 딕셔너리
    """
    network = {}
    
    # 은닉층 1 가중치와 편향
    network['W1'] = np.array([[0.1, 0.3, 0.5],
                              [0.2, 0.4, 0.6]])
    network['b1'] = np.array([0.1, 0.2, 0.3])
    
    # 은닉층 2 가중치와 편향
    network['W2'] = np.array([[0.1, 0.4],
                              [0.2, 0.5],
                              [0.3, 0.6]])
    network['b2'] = np.array([0.1, 0.2])
    
    # 출력층 가중치와 편향
    network['W3'] = np.array([[0.2, 0.5],
                              [0.3, 0.6]])
    network['b3'] = np.array([0.1, 0.2])
    
    return network


# 초기화된 신경망 확인
network = init_network()
print("초기화된 신경망:")
for key, value in network.items():
    print(f"  {key}: 형태 = {value.shape}")
````

````python
def forward(network, x):
    """
    순전파 (Forward Propagation) 함수
    
    입력 데이터 x를 신경망에 입력하여 출력을 계산합니다.
    
    Parameters
    ----------
    network : dict
        초기화된 신경망 (가중치와 편향 포함)
    x : numpy array
        입력 데이터
    
    Returns
    -------
    numpy array: 출력층의 예측값 (softmax 적용 후)
    """
    # 가중치와 편향 추출
    W1, W2, W3 = network['W1'], network['W2'], network['W3']
    b1, b2, b3 = network['b1'], network['b2'], network['b3']
    
    # 은닉층 1: 가중합 계산 → 시그모이드 활성화
    a1 = np.dot(x, W1) + b1
    z1 = sigmoid(a1)
    
    # 은닉층 2: 가중합 계산 → 시그모이드 활성화
    a2 = np.dot(z1, W2) + b2
    z2 = sigmoid(a2)
    
    # 출력층: 가중합 계산 → Softmax 활성화
    a3 = np.dot(z2, W3) + b3
    y = softmax(a3)
    
    return y


# 순전파 함수 테스트
X = np.array([1.0, 0.5])
output = forward(network, X)

print(f"입력: {X}")
print(f"출력: {output}")
print(f"출력값 합계: {np.sum(output)}")
print(f"예측 클래스: {np.argmax(output)}")
````

## 다층 퍼셉트론 (MLP) 요약

### 신경망의 구성 요소

| 요소 | 설명 |
|------|------|
| **입력층** | 외부 데이터를 받아들이는 층 (이 예제에서는 2개 뉴런) |
| **은닉층** | 입력과 출력 사이에 있는 층 (여러 개 가능) |
| **출력층** | 최종 결과를 출력하는 층 (이 예제에서는 2개 뉴런) |
| **가중치 (W)** | 뉴런 간 연결의 강도, 학습을 통해 조정 |
| **편향 (b)** | 뉴런의 활성화 임계값 |
| **활성화 함수** | 비선형 변환을 적용하여 복잡한 패턴 학습 가능 |

### 순전파 과정

```
입력 X → [은닉층 1: A1=W1X+b1, Z1=sigmoid(A1)] → [은닉층 2: A2=W2Z1+b2, Z2=sigmoid(A2)] → [출력층: A3=W3Z2+b3, Y=softmax(A3)] → 출력 Y
```

### 활성화 함수 선택

| 층 | 활성화 함수 | 이유 |
|----|------------|------|
| 은닉층 1 | sigmoid | 입력을 0~1 사이로 압축 |
| 은닉층 2 | sigmoid | 입력을 0~1 사이로 압축 |
| 출력층 | softmax | 확률 분포로 변환 (합계=1) |

### 주요 개념

- **순전파 (Forward Propagation)**: 입력에서 출력 방향으로 신호가 전달되는 과정
- **가중합**: 입력 × 가중치 + 편향
- **비선형성**: 활성화 함수를 통해 신경망이 복잡한 결정 경계 학습 가능
- **Softmax**: 다중 분류 문제에서 확률 출력으로 변환
