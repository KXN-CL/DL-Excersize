```python
# coding: utf-8
# ReLU 함수 (ReLU Function) 구현
# ReLU (Rectified Linear Unit) 함수는 신경망에서 가장 널리 사용되는 활성화 함수로,
# 입력이 양수이면 해당 값을 출력하고, 음수이면 0을 출력합니다.
# 수식: f(x) = x (x > 0일 때), f(x) = 0 (x <= 0일 때)

import numpy as np
import matplotlib.pylab as plt


def relu(x):
    """
    ReLU (Rectified Linear Unit) 함수.
    
    이 함수는 입력 x가 0보다 크면 x 값을 반환하고,
    0보다 작거나 같으면 0을 반환합니다.
    
    Args:
        x: 입력 값 - 스칼라, 리스트, 또는 numpy 배열 가능
        
    Returns:
        np.ndarray: ReLU 함수를 적용한 출력 값
    """
    # numpy의 maximum 함수를 사용하여 x와 0 중 큰 값을 선택합니다
    # x > 0 이면 x 를, x <= 0 이면 0 을 반환합니다
    return np.maximum(0, x)


# -7.0 에서 7.0 까지 0.1 간격으로 입력 값 생성
# ReLU 함수의 선형/영역 구분을 명확히 보기 위한 범위입니다
X = np.arange(-7.0, 7.0, 0.1)

# 모든 입력 값에 ReLU 함수를 적용합니다
Y = relu(X)

# ReLU 함수를 그래프로 plotting 합니다
# X 는 입력 값을 포함하고, Y 는 해당하는 출력 값을 포함합니다
plt.plot(X, Y)

# 그래프 설정
plt.title('ReLU Function', fontsize=16)
plt.xlabel('Input (x)', fontsize=12)
plt.ylabel('Output (relu(x))', fontsize=12)
plt.grid(True, alpha=0.3)

# y 축 범위를 설정합니다
plt.ylim(-1, 8)  # y 축의 범위 설정

# x = 0 위치에 점선 표시 (ReLU 함수가 변화하는 지점)
plt.axvline(x=0, color='red', linestyle='--', alpha=0.5)
# y = 0 위치에 점선 표시
plt.axhline(y=0, color='green', linestyle='--', alpha=0.5)

# 그래프를 표시합니다
plt.show()
```
