```python
# coding: utf-8
# OR 게이트 구현: 두 입력 중 하나 이상 1이면 1을 출력하는 논리 게이트
import numpy as np


def OR(x1, x2):
    """
    OR 게이트 함수
    
    입력:
        x1: 첫 번째 입력 (0 또는 1)
        x2: 두 번째 입력 (0 또는 1)
    
    출력:
        0 또는 1 (OR 연산 결과)
    
    동작 원리:
        단순 퍼셉트론을 사용하며, 다음 수식으로 표현됨:
        y = sum(w * x) + b
        여기서 w = [0.5, 0.5], b = -0.2
        sum(w * x) + b >= 0 이면 1, 아니면 0을 반환
        AND 게이트(b=-0.7)보다 편향 값이 작아(더 작음) 한 개의 입력만 있어도 1이 출력됨
    """
    # 입력 값을 numpy 배열로 변환
    x = np.array([x1, x2])
    # 가중치 설정: 두 입력에 동일한 가중치 0.5 할당
    w = np.array([0.5, 0.5])
    # 편향 설정: -0.2로 AND 게이트보다 작은 음수 값
    b = -0.2
    # 가중치와 입력의 곱의 합에 편향을 더함 (내적 계산)
    tmp = np.sum(w*x) + b
    # 합이 0 이하이면 0, 아니면 1을 반환 (단계 함수 활성화 함수 사용)
    if tmp <= 0:
        return 0
    else:
        return 1

if __name__ == '__main__':
    # OR 게이트의 모든 경우의 수(4가지)에 대해 테스트
    for xs in [(0, 0), (1, 0), (0, 1), (1, 1)]:
        y = OR(xs[0], xs[1])
        print(str(xs) + " -> " + str(y))
```
