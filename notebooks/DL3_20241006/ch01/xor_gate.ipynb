```python
# coding: utf-8
# XOR 게이트 구현: 두 입력이 서로 다르면 1, 같으면 0을 출력하는 논리 게이트
# XOR 게이트는 단일 퍼셉트론으로 구현할 수 없으며, 다층 퍼셉트론(Multi-Layered Perceptron)이 필요함
# NAND 게이트, OR 게이트를 조합하여 구현
import numpy as np


def NAND(x1, x2):
    """
    NAND 게이트 함수
    
    입력:
        x1: 첫 번째 입력 (0 또는 1)
        x2: 두 번째 입력 (0 또는 1)
    
    출력:
        0 또는 1 (NAND 연산 결과)
    """
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.9  # NAND 게이트의 편향 값
    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1


def OR(x1, x2):
    """
    OR 게이트 함수
    
    입력:
        x1: 첫 번째 입력 (0 또는 1)
        x2: 두 번째 입력 (0 또는 1)
    
    출력:
        0 또는 1 (OR 연산 결과)
    """
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.2  # OR 게이트의 편향 값
    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1


def XOR(x1, x2):
    """
    XOR 게이트 함수 (다층 퍼셉트론 사용)
    
    입력:
        x1: 첫 번째 입력 (0 또는 1)
        x2: 두 번째 입력 (0 또는 1)
    
    출력:
        0 또는 1 (XOR 연산 결과)
    
    동작 원리:
        XOR 게이트는 단일 퍼셉트론으로 구현할 수 없는 선형 비분리 문제(linearly inseparable problem)
        따라서 NAND 게이트와 OR 게이트를 조합한 다층 퍼셉트론으로 구현:
        1. 제1층: s1 = NAND(x1, x2), s2 = OR(x1, x2)
        2. 제2층: y = AND(s1, s2)
        수식으로 표현: y = AND(NAND(x1, x2), OR(x1, x2))
    """
    # 첫 번째 층: NAND와 OR 게이트 계산
    s1 = NAND(x1, x2)
    s2 = OR(x1, x2)
    # 두 번째 층: AND 게이트로 s1과 s2를 입력받아 최종 출력 계산
    y = AND(s1, s2)
    return y


def AND(x1, x2):
    """
    AND 게이트 함수 (XOR 구현에 사용)
    
    입력:
        x1: 첫 번째 입력 (0 또는 1)
        x2: 두 번째 입력 (0 또는 1)
    
    출력:
        0 또는 1 (AND 연산 결과)
    """
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7  # AND 게이트의 편향 값
    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1

if __name__ == '__main__':
    # XOR 게이트의 모든 경우의 수(4가지)에 대해 테스트
    for xs in [(0, 0), (1, 0), (0, 1), (1, 1)]:
        y = XOR(xs[0], xs[1])
        print(str(xs) + " -> " + str(y))
```
