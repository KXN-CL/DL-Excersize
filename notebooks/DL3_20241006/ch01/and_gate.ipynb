```python
# coding: utf-8
# AND 게이트 구현: 두 입력이 모두 1일 때만 1을 출력하는 논리 게이트
import numpy as np


def AND(x1, x2):
    """
    AND 게이트 함수
    
    입력:
        x1: 첫 번째 입력 (0 또는 1)
        x2: 두 번째 입력 (0 또는 1)
    
    출력:
        0 또는 1 (AND 연산 결과)
    
    동작 원리:
        단순 퍼셉트론(Simple Perceptron)을 사용하며, 다음 수식으로 표현됨:
        y = sum(w * x) + b
        여기서 w = [0.5, 0.5], b = -0.7
        sum(w * x) + b >= 0 이면 1, 아니면 0을 반환
    """
    # 입력 값을 numpy 배열로 변환
    x = np.array([x1, x2])
    # 가중치(wweight) 설정: 두 입력에 동일한 가중치 0.5 할당
    w = np.array([0.5, 0.5])
    # 편향(bias) 설정: -0.7로 설정하여 입력이 없을 때는 0을 출력하도록 함
    b = -0.7
    # 가중치와 입력의 곱의 합에 편향을 더함 (내적 계산)
    tmp = np.sum(w*x) + b
    # 합이 0 이하이면 0, 아니면 1을 반환 (단계 함수 활성화 함수 사용)
    if tmp <= 0:
        return 0
    else:
        return 1

if __name__ == '__main__':
    # AND 게이트의 모든 경우의 수(4가지)에 대해 테스트
    for xs in [(0, 0), (1, 0), (0, 1), (1, 1)]:
        y = AND(xs[0], xs[1])
        print(str(xs) + " -> " + str(y))
```
