```python
# coding: utf-8
# NAND 게이트 구현: AND 게이트의 역으로, 두 입력이 모두 1일 때만 0을 출력
# NAND 게이트는 AND 게이트보다 편향(bias) 값을 조정하여 구현 가능
import numpy as np


def NAND(x1, x2):
    """
    NAND 게이트 함수
    
    입력:
        x1: 첫 번째 입력 (0 또는 1)
        x2: 두 번째 입력 (0 또는 1)
    
    출력:
        0 또는 1 (NAND 연산 결과)
    
    동작 원리:
        NAND는 NOT(AND)로, AND 게이트의 출력을 반전시킨다.
        가중치 w = [0.5, 0.5], 편향 b = -0.9로 설정
        AND 게이트(b=-0.7)보다 더 큰 음수 편향을 주어, 두 입력이 모두 1일 때만 0이 출력되도록 함
    """
    # 입력 값을 numpy 배열로 변환
    x = np.array([x1, x2])
    # 가중치 설정: AND 게이트와 동일한 가중치 사용
    w = np.array([0.5, 0.5])
    # 편향 설정: -0.9로 AND 게이트(-0.7)보다 더 큰 음수 값
    b = -0.9
    # 가중치와 입력의 곱의 합에 편향을 더함
    tmp = np.sum(w*x) + b
    # 합이 0 이하이면 0, 아니면 1을 반환
    if tmp <= 0:
        return 0
    else:
        return 1

if __name__ == '__main__':
    # NAND 게이트의 모든 경우의 수(4가지)에 대해 테스트
    for xs in [(0, 0), (1, 0), (0, 1), (1, 1)]:
        y = NAND(xs[0], xs[1])
        print(str(xs) + "->" + str(y))
```
